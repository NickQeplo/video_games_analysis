Перед заменой пропусков стоит посмотреть однородность данных. Изучить максимальные и минимальные значения столбцов: eu_sales, jp_sales, critic_score, user_score.

Максимальные значения полученные в столбцах eu_sales, jp_sales сильно выше, чем 3 квантиль, поэтому стоит заменить пропуски на медианное значение. В том время как в столбцах critic_score, user_score данные более однородные, здесь пропуски следует заменить на среднее значение.

В ходе первоначального решение, некоторые данные после группировки ( год издания и платформа ) не имели вообще никакой оценки. Поэтому для таких групп пропуски все равно останустя. Таких строк всего 1598, что составляет почти 10% всех данных, поэтому  данные значения стоит заменить на среднее значение по платформе. 

```python
grouped_platforms = games.groupby('platform')[['user_score', 'critic_score']]

grouped_platforms.describe()
```

Из результатов видно, что не группировка данных по  году и платформе дает отсутсвие данных, а у некоторых платформ отсутствуют оценки, поэтому стоит рассмотреть решение заполнить пропуски по среднему значению жанра.


```python
grouped_platforms = games.groupby('genre')[['user_score', 'critic_score']]

grouped_platforms.describe()
```

В результате видны неявные дубликаты, поэтому стоит сначала обработать дубликаты, а потом посмотреть однородность данных и заменить пропуски, поэтомум сейчас заполним только столбцы eu_sales, jp_sales.

```python
columns = ['eu_sales', 'jp_sales']

for column in columns:  
    median = games.groupby(['platform', 'year_of_release'])[column].transform('median')
    games[column].fillna(value=median, inplace=True)
# Проверка измененных значений
for column in columns:  
        print(f'Кол-во пропусков в столбце {column}: {(games[column].isna()).sum()}')
```

###  Явные и неявные дубликаты в данных

Рассмотрим уникальные значения в категориальных столбцах - name, platform, year_of_release, genre

```python
columns = ['platform', 'year_of_release', 'genre']

for column in columns:
    print(f'Уникальные значения столбца {column}: ', games[column].unique(), '\n')
```

Данные в столбцах platform и year_of_release приведены корректно. В столбце genre встречаются неявные дубликаты: некоторые жанры написаны заглавными буквами, приведем все названия жанров к нижнему регистру.

```python
games['genre'] = games['genre'].str.lower()
# Проверка
games['genre'].unique()
```

Данные нормализованы, теперь можно рассмотреть возможность замены оценок по жанру игры.

```python
grouped_platforms = games.groupby('genre')[['user_score', 'critic_score']]

grouped_platforms.describe()
```

Теперь у каждой группы есть оценки, данные выглядят однородными. Пропуски можно заменить на среднее значение по жанру.

```python
columns = ['user_score', 'critic_score']

for column in columns:
    mean_of_genre = games.groupby('genre')[column].transform('mean')
    games[column] = games[column].fillna(mean_of_genre)

# Проверка
for column in columns:
    print(f'Кол-во пропусков в {column}: {games[column].isna().sum()}')
```

```python
# стоит проверить дубликаты по названию игры и платформе
games.duplicated(subset=['name', 'platform']).sum()
```

Результат 238, а значит есть строки, которые описывают одну и ту же игру.

```python
games[games.duplicated(subset=['name', 'platform'])]
```

Принято решение удалить повторяющиеся строки.

```python
games.drop_duplicates(subset=['name', 'platform'], inplace=True)
```

## Фильтрация данных

```python
# Так нужен срез датафрейма, использую метод copy(),
# иначе df_actual будет хранить в себе лишь ссылку на исходны датафрейм games,
# и pandas будет предупреждать о люом измненение в df_actual
df_actual = games[games['year_of_release'].between(2000,2013)].copy()
df_actual.info()
```

Для дальнейшего анализа был выделен актуальный период с 2000 по 2013 год включительно. После фильтрации в датафрейме осталось 12778 записей.

## Категоризация данных
    
- Разделил все игры по оценкам пользователей: высокая оценка (от 8 до 10 включительно), средняя оценка (от 3 до 8, не включая правую границу интервала) и низкая оценка (от 0 до 3, не включая правую границу интервала).

```python
categories = ['низкая оценка', 'средняя оценка', 'высокая оценка']

# Расширил верхнюю границу, чтобы игры, где оценка 10 тоже получили категорию
# Предварительно были рассмотренны максимальные значения оценок, где макс составлял 9.7.
df_actual['category_by_user'] = pd.cut(df_actual['user_score'], bins=[0, 3, 8, 10.1], labels=categories, right=False, include_lowest=True)

# Проверка
df_actual[['user_score', 'category_by_user']]
```

- Похожим образом разделил игры по оценкам критиков: высокая оценка (от 80 до 100 включительно), средняя оценка (от 30 до 80, не включая правую границу интервала) и низкая оценка (от 0 до 30, не включая правую границу интервала).

```python
# Расширил верхнюю границу, чтобы игры, где оценка 100 тоже получили категорию
# Предварительно были рассмотренны максимальные значения оценок, где макс составлял 98.
df_actual['category_by_critic'] = pd.cut(df_actual['critic_score'], bins=[0, 30, 80, 100.1], labels=categories, right=False, include_lowest=True)

# Проверка
df_actual[['critic_score', 'category_by_critic']]
```

```python
df_actual.groupby('category_by_user')['name'].count()
```

```python
df_actual.groupby('category_by_critic')['name'].count()
```

```python
df_actual.groupby(['category_by_user', 'category_by_critic'])['name'].count()
```

- Выделил топ-7 платформ по количеству игр, выпущенных за весь актуальный период.

```python
df_actual.groupby('platform')['name'].count().sort_values(ascending=False)[:7]
```

Игры были разделены на категории по пользовательским оценкам и по оценкам критиков:
- низкая оценка
- средняя оценка
- высокая оценка

Были определены наиболее популярные платформы по количеству выпущенных игр за актуальный период, выделен топ-7 платформ.
